# Train on Kaggle GPU

**One-time setup:**
1. Attach the dataset: right panel → *Input* → *Add input* → your `mesogeos-track-a-processed`.
2. Enable GPU: right panel → *Session options* → Accelerator → GPU P100/T4.
3. GitHub access (repo is private): create a fine-grained PAT at github.com/settings/personal-access-tokens
   (read-only, only the `wildfire-dissertation` repo), then here: *Add-ons → Secrets → Add secret*,
   name it `GITHUB_PAT`.

Then: Run all. Change `CONFIG` to train a different model.
